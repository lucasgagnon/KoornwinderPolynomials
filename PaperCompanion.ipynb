# Section 4

```python
# The formulas in this section provide TWO ways to represent Koornwinder polynomials
# First, as laurent polynomials in a KoornwinderPolynomialRing 

R = KoornwinderPolynomialRing(3)
K = R.base_ring()

f = x_1 * x_2 + sqrt_q * x_1
show(f)
```

```python
# OR in a new `C/F function space' which represents C and F functions abstractly, at the cost of
# non-uniqueness of representation.

CRing = CF_function_space(3, QQ)

# Elements of this space are formal monomials indexed by weights, a choice of C/F+/F-, and exponent

C_monomials = CF_indices(3)
mon1 = C_monomials({(1, 0, 0, -1/2) : {'C' : 2}}) # C_{-alpha_0} ^ 2
mon2 = C_monomials({(-1, 1, 0, 0) : {'F+' : 1}})  # F+_{-alpha_1}

f = 3 * sqrt_t * CRing(mon1) + x_1 * CRing(mon2)
show(f)

# We can reduce these elements to ordinary polynomials using the .expand() command.

f.expand()
```

# The creation formula

```python
# The tau/creation operations are built into both realizations, but 
# require the Macdonald eigenvector being acted on in order to be computable
```

```python
# For laurent polynomials

E100 = 1/(sqrt_t^4 * sqrt_tn) * Laurent_tau(0, [0, 0, 0], R(1))
show(E100)
```

```python
# For CF-space elements

CF_E000 = CRing(1)

CF_E100 = 1/(sqrt_t^4 * sqrt_tn) * CF_E000.tau(0, [0, 0, 0])
show(CF_E100)

# After `expanding' this produces the same result

CF_E100.expand() == E100
```

# Section 4.2: The alcove walk formula

```python
# We can represent alcove walks as a class which reads sequences of UNCROSSED indices
# as booleans and an initial signed permutation z

mu = KoornwinderBoxDiagram([0, 2, 3, -1, 1])
T = AlcoveWalkTableaux(mu, [i+1 not in [3, 6, 8, 15, 17, 22, 26, 32, 37] for i in range(len(mu.box_greedy_reduced_word()))], z = [1, 2, 3, 4, 5])
T.pretty_print()
print(len(T))
```

```python
# The permutation sequence of a box can be accessed as a method of the alcove walk
T.permutation_sequence(2, 3)
```

```python
# The roow sequence of a box can also be accessed as a method of the alcove walk
T.root_sequence(2, 3)
```

```python
# One can also inspect the weight of an individual box
# This returns an F-function, which can be .expand()-ed to a Laurent series
show(AlcoveWalkBoxWeight(T, 2, 3))
```

```python
# the function AlcoveWalks returns all alcove walks for a given shape and z:

for AW in AlcoveWalks([0, 0, 1], z = None):
    print("Folding =", str(AW))
    print("Permutation sequence =", str(AW.full_permutation_sequence()))
    AW.pretty_print()
    print(AlcoveWalkWeight(AW))
    print()
    print()
```
